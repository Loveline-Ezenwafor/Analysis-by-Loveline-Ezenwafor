### Pandas - Hands-on Exercises

This notebook contains simple exercises in Pandas. All the exercises will be based on the "Introduction to Pandas - Beginner's Guide" Jupyter notebook.

There may be one or more ways to find solution to a particular exercise, so try to experiment most in order to get used to the dataframes and series.

```python
#import pandas
import pandas as pd
```

```python
# create a pandas Series object with different fruit names.
fruits = pd.Series(['Apple', 'Banana', 'Cherry', 'Pineapple', 'Watermelon'])
```

```python
#View the series of different fruits
# Print the Series
print(fruits)
```

```python
#create a sample dataframe
data = {
    'Fruit': ['Apple', 'Banana', 'Cherry', 'Pineapple', 'Watermelon'],
    'Quantity': [10, 20, 15, 5, 8],
    'Price': [1.2, 0.5, 2.0, 3.0, 1.5]
}
df = pd.DataFrame(data)
print("\nSample DataFrame:")
print(df)
```

```python
#Read a CSV file into a pandas DataFrame.
df.to_csv('data.csv', index=False)

```

```python
#Display the top 5 rows in the dataframe
print("\nTop 5 rows:")
print(df.head())
```

```python
#Display the bottom 5 rows in the dataframe
print("\nBottom 5 rows:")
print(df.tail())
```

```python
#select first 10 columns using loc function
# Assuming the DataFrame has at least 10 columns
# Note: You need to adjust '10' based on the actual number of columns in your DataFrame
print("\nFirst 10 columns:")
print(df.loc[:, df.columns[:10]])
```

```python
# retrieve the shape of the dataframe
print("\nShape of the DataFrame:")
print(df.shape)
```

```python
# find the null values count.
print("\nNull values count:")
print(df.isnull().sum())
```

```python
# filter rows based on a condition
# Example: Filter rows where Quantity > 10
filtered_df = df[df['Quantity'] > 10]
print("\nFiltered rows (Quantity > 10):")
print(filtered_df)
```

```python
#sort the dataframe in ascending order
# Example: Sort by 'Price'
sorted_df = df.sort_values(by='Price')
print("\nSorted DataFrame by Price:")
print(sorted_df)
```

```python
#Find the sum and mean of any of the numerical column present in the dataframe.
# Example: Find sum and mean of 'Quantity'
quantity_sum = df['Quantity'].sum()
quantity_mean = df['Quantity'].mean()

print("\nSum of Quantity:", quantity_sum)
print("Mean of Quantity:", quantity_mean)
```

```python
#groupby the dataframe using a columns and display the first records.
# Example: Group by 'Fruit' and display the first record in each group
grouped_df = df.groupby('Fruit').first()
print("\nGrouped by Fruit (first record in each group):")
print(grouped_df)
```
